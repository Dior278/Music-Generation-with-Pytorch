```python
from transformers import pipeline

summerize = pipeline("summarization", model="facebook/bart-large-cnn")

text = """Transformers is an open-source library developed by Hugging 
        Face that provides state-of-the-art machine learning models 
        primarily focused on natural language processing (NLP) tasks. 
        It offers a wide range of pre-trained models, including BERT, 
        GPT, T5, and many others, which can be easily fine-tuned for 
        various applications such as text classification, translation, 
        summarization, and question answering. """   

summary = summerize(text, max_length=50)
print(summary[0]['summary_text'])
```
